### De modelos de linguagem a grandes modelos de linguagem · PDF p. 42

#### Exercício: Ver os tokens de uma frase

O livro explica que modelos leem **tokens**, não palavras. Escolha uma frase em português e a mesma frase em inglês. Use `contar_tokens()` para comparar quantos tokens cada uma tem e, com o `tiktoken`, mostre cada token separado.

Dica: `tiktoken.get_encoding('o200k_base')` → `.encode(texto)` devolve números; `.decode([n])` mostra o pedaço de texto de cada número.

```python
import tiktoken
enc = tiktoken.get_encoding("o200k_base")

frases = {
    "pt": "Meu nome é Raphael, tenho 43 anos",
    "en": "My name is Raphael, I'm 43 years old",
}
for idioma, frase in frases.items():
    ids = enc.encode(frase)
    pedacos = [enc.decode([i]) for i in ids]
    print(f"{idioma}: {len(ids)} tokens -> {pedacos}")
```

## Casos de uso do modelo Foundation · PDF p. 66

#### Exercício: Sua primeira chamada: classificar mensagens

Monte uma lista com 5 mensagens de clientes (reclamação, elogio, dúvida…). Use `perguntar()` para classificar cada uma em **reclamação / elogio / dúvida / outro**, pedindo que o modelo responda **só a categoria**. Mostre o resultado numa tabela (pandas).

```python
import pandas as pd
mensagens = [
    "O pedido chegou com a caixa amassada.",
    "Adorei o atendimento, foram muito rápidos!",
    "Vocês entregam no sábado?",
    "Quero cancelar minha assinatura.",
    "Qual o prazo de garantia do produto?",
]
sistema = ("Classifique a mensagem do cliente em exatamente uma categoria: "
           "reclamação, elogio, dúvida ou outro. Responda só com a categoria, em minúsculas.")
resultado = [{"mensagem": m, "categoria": perguntar(m, sistema=sistema, temperature=0)} for m in mensagens]
pd.DataFrame(resultado)
```
